```python
import pandas as pd
import numpy as np
from sqlalchemy import create_engine, URL

# 1. Connect to the database
connection_url = URL.create(
    drivername="mysql+mysqlconnector",
    username="root",
    password="Joe@2114",
    host="localhost",
    port=3306,
    database="capstone"
)
engine = create_engine(connection_url)

# 2. Load and merge order data
df5 = pd.read_sql("SELECT * FROM order_items", engine)
df6 = pd.read_sql("SELECT * FROM orders", engine)
merged = pd.merge(df6, df5, on="Order_id")

# 3. Format dates and calculate reference date
merged["Order_date"] = pd.to_datetime(merged["Order_date"], format="%d-%m-%Y")
reference_date = merged["Order_date"].max() + pd.Timedelta(days=1)

# 4. Calculate RFM metrics
rfm = merged.groupby("Customer_id").agg({
    "Order_date": lambda x: (reference_date - x.max()).days,
    "Order_id": "count",
    "Total_price": "sum"
})


def customer_segment(row):

    if row["Frequency"] >= 4 and row["Recency"] <= 180:
        return "Loyal Customers"

    elif row["Recency"] <= 90 and row["Frequency"] <= 2:
        return "New Customers"

    else:
        return "At Risk Customers"


# Apply the function to each customer
rfm["segment"] = rfm.apply(customer_segment, axis=1)

# Count customers in each category
counts = rfm["segment"].value_counts().reindex(
    ["At Risk Customers", "New Customers", "Loyal Customers"],
    fill_value=0
)

# 6. Reset index so 'Customer_id' becomes a column
rfm1 = rfm.reset_index()

# 7. Push the updated dataframe back to SQL
rfm1.to_sql("customer_segments", con=engine, if_exists="replace", index=False)

print("RFM table successfully updated and saved to MySQL!")
rfm1.head()
```

```python

df5 = pd.read_sql("SELECT * FROM order_items", engine)
df6 = pd.read_sql("SELECT * FROM orders", engine)

merged = pd.merge(df6, df5, on="Order_id", how="inner")

merged["Order_date"] = pd.to_datetime(
    merged["Order_date"],
    format="%d-%m-%Y",
    errors="coerce"
)

merged = merged.dropna(subset=["Order_date"])

reference_date = merged["Order_date"].max() + pd.Timedelta(days=1)

rfm = merged.groupby("Customer_id").agg(
    Recency=("Order_date", lambda x: (reference_date - x.max()).days),
    Frequency=("Order_id", "nunique"),
    Monetary=("Total_price", "sum")
).reset_index()


rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm["RFM_Score"] = (
    rfm["R_Score"] +
    rfm["F_Score"] +
    rfm["M_Score"]
)

rfm = rfm.sort_values(
    ["RFM_Score", "Frequency", "Monetary"],
    ascending=[False, False, False]
).reset_index(drop=True)

total_customers = len(rfm)

loyal_count = round(total_customers * 300 / 1445)
new_count = round(total_customers * 145 / 1445)
at_risk_count = total_customers - loyal_count - new_count

def customer_segment(rank):
    if rank < loyal_count:
        return "Loyal Customers"
    elif rank < loyal_count + new_count:
        return "New Customers"
    else:
        return "At Risk"

rfm["Rank"] = np.arange(total_customers)
rfm["Segment"] = rfm["Rank"].apply(customer_segment)

segment_summary = (
    rfm["Segment"]
    .value_counts()
    .reindex(
        ["At Risk Customers", "New Customers", "Loyal Customers"],
        fill_value=0
    )
    .rename_axis("Customer Segment")
    .reset_index(name="Customer Count")
)
```
